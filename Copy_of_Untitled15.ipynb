```python
# ==================== Node ====================
class Node:
    def __init__(self, value):
        self.left = None
        self.data = value
        self.right = None


# ==================== Tree ====================
class Tree:

    # ---------- Create & Insert ----------
    def createNode(self, data):
        return Node(data)

    def insert(self, node, data):
        if node is None:
            return self.createNode(data)

        if data < node.data:
            if node.left is None:
                print(f"Inserted {data} -> left of {node.data}")
            node.left = self.insert(node.left, data)
        else:
            if node.right is None:
                print(f"Inserted {data} -> right of {node.data}")
            node.right = self.insert(node.right, data)

        return node

    # ---------- Traversals ----------
    def traverse_Inorder(self, root):
        if root is not None:
            self.traverse_Inorder(root.left)
            print(root.data, end=" ")
            self.traverse_Inorder(root.right)

    def traverse_Preorder(self, root):
        if root is not None:
            print(root.data, end=" ")
            self.traverse_Preorder(root.left)
            self.traverse_Preorder(root.right)

    def traverse_Postorder(self, root):
        if root is not None:
            self.traverse_Postorder(root.left)
            self.traverse_Postorder(root.right)
            print(root.data, end=" ")


# ==================== Driver Code ====================
tree = Tree()
root = tree.createNode(5)
print(f"Root created: {root.data}")

print("\n--- Insert ---")
for value in [2, 10, 7, 15, 12, 20, 30, 6, 8]:
    tree.insert(root, value)

print("\n--- Traversals ---")
print("Inorder   -->", end=" ")
tree.traverse_Inorder(root)

print("\nPreorder  -->", end=" ")
tree.traverse_Preorder(root)

print("\nPostorder -->", end=" ")
tree.traverse_Postorder(root)
print()
```
